###  Now we create database for years 2000-2025

```python


url = "https://f1-motorsport-data.p.rapidapi.com/schedule"
 
headers = {
	"x-rapidapi-key": "cc30b177f7msh2b0ba96cfd0d061p10144ejsna3cc2725a075",
	"x-rapidapi-host": "f1-motorsport-data.p.rapidapi.com",
	"Content-Type": "application/json"
}

all_dfs = []

for year in range(2000, 2026):
    querystring = {"year": year}
    response = requests.get(url, headers=headers, params=querystring)

    if response.status_code == 200:
        data = response.json()

        df = pd.json_normalize(data)[[
            'startDate',
            'endDate',
            'gPrx',
            'crct',
            'winnerAthlete.id',
            'winnerAthlete.name',
            'winnerAthlete.flag',
            'winnerAthlete.citizenship',
            'evLink',
            'completed'
        ]]

        df = df.rename(columns={
            'startDate': 'Start Date',
            'endDate': 'End Date',
            'gPrx': 'Grand Prix',
            'crct': 'Circuit',
            'winnerAthlete.id': 'Winner ID',
            'winnerAthlete.name': 'Winner',
            'winnerAthlete.flag': 'Flag',
            'winnerAthlete.citizenship': 'Country',
            'evLink': 'Event Link',
            'completed': 'Completed'
        })

        all_dfs.append(df)  #pehla dataframe bnao per year, and usko ek ek krke add kro
                            #option 2 -> pehle per year ka json bnao ,  and combinely saare json ko last me dataframe me convert krdo

    else:
        print(f"{year}: {response.status_code}")

    time.sleep(1)    # Avoid 429 rate-limit errors

# Combine all years into one DataFrame
final_df = pd.concat(all_dfs, ignore_index=True)

display(final_df)
```
